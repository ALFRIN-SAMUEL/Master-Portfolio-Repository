```python
import pandas as pd 
import zipfile
name="online+retail.zip"
with zipfile.ZipFile(name,'r') as asww:
     m=asww.namelist()
     print(m)
     with asww.open(m[0]) as op:
        df=pd.read_excel(op)
df.head()

     
```

```python
import pymysql
from sqlalchemy import create_engine
engine=create_engine('mysql+pymysql://root:password@localhost:3306/sales_db')
df.to_sql("rawsales",con=engine,if_exists='replace',index=False)
```
